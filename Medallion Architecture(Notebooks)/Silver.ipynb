```python
bronze_supply_path = "abfss://fbronze@storagedatafactory2605.dfs.core.windows.net/supply_chain_delta/"

df_supply_bronze = spark.read.format("delta").load(bronze_supply_path)

display(df_supply_bronze)
df_supply_bronze.printSchema()
df_supply_bronze.count()
```

```python
from pyspark.sql import functions as F

df_supply = df_supply_bronze

# Trim + Lowercase for important text columns
df_supply = (
    df_supply
    .withColumn("product_name", F.lower(F.trim(F.col("product_name"))))
    .withColumn("category_name", F.lower(F.trim(F.col("category_name"))))
    .withColumn("market", F.lower(F.trim(F.col("market"))))
    .withColumn("order_status", F.lower(F.trim(F.col("order_status"))))
)
```

```python
df_supply = (
    df_supply
    .withColumn("order_ts", F.to_timestamp("order_date_dateorders", "yyyy-MM-dd HH:mm:ss"))
    .withColumn("ship_ts", F.to_timestamp("shipping_date_dateorders", "yyyy-MM-dd HH:mm:ss"))
)
```

```python
df_supply = (
    df_supply
    .withColumn("order_year", F.year("order_ts"))
    .withColumn("order_month", F.month("order_ts"))
    .withColumn("order_ym", F.date_format("order_ts", "yyyy-MM"))
)
```

```python
api_path = "abfss://destination@storagedatafactory2605.dfs.core.windows.net"

df_api_bronze = (
    spark.read.option("header", "true").option("inferSchema", "true").csv(api_path)
)

display(df_api_bronze)
df_api_bronze.printSchema()
df_api_bronze.count()
```

```python
from pyspark.sql import functions as F

df_api_silver = (
    df_api_bronze
    # Clean DESCRIPTION column
    .withColumn(
        "description",
        F.trim(F.regexp_replace("DESCRIPTION", r"^:|^:\s*", ""))  # remove ":" at start
    )
    # Clean FIELDS column
    .withColumn(
        "field_name",
        F.trim(F.regexp_replace("FIELDS", r"[\s:]+$", ""))  # remove trailing space or ':' 
    )
    # Final cleanup: lowercase + underscores for Delta-safety
)

# Standardize final column names (safe for Delta)
df_api_silver = df_api_silver.select(
    F.col("description"),
    F.lower(F.regexp_replace("field_name", r"[^A-Za-z0-9]+", "_")).alias("field_name")
)

# Remove duplicates & nulls just in case
df_api_silver = df_api_silver.dropDuplicates().filter("description IS NOT NULL")

display(df_api_silver)
df_api_silver.printSchema()
print(df_api_silver.count())
```

```python
api_silver_path = "abfss://fsilver@storagedatafactory2605.dfs.core.windows.net/api_docs_delta/"

df_api_silver.write.format("delta").mode("overwrite").save(api_silver_path)
```
